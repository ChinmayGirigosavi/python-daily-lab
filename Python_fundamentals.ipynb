6. Identity Operators (is & is not)
* The 'is' and 'is not' operator in Python checks whether
two variables refer to the exact same object in memory.
* It is known as an identity operator.

```python
# both are used to check if two values are located on same part of the memory

a = 20
b = 25
print(a is not b)
print(a is b)

b = a
print(a is b)

# is -> True if the operands are identical
# is not -> True if the operands are not identical
```

7. [Membership Operator](https://www.geeksforgeeks.org/python/python-membership-identity-operators-not-not/)

* 'in' is a membership operator in Python.
* It is used to test whether a value or an element exists within a sequence or a collection (such as a string, list, tuple, set, or dictionary)

```python
# in and not in are the membership operators
rollno_list = [23,24, 25, 26, 28]
subject = "Foundation of Data Science"
dictionary = {1: "Foundation", 2: "of", 3: "Data Science"}

print(28 in rollno_list)
print(30 in rollno_list)
# returns Flase because the element 30 does not exists inside the given sequence (rollno_list)

print(30 not in rollno_list, "\n")
# 'not in' operator confirms that the element 30 does not exists in rollno_list (given sequence)


print("Foundation" in subject)
print("of" in subject)
print("Of" in subject, "\n")
# returns False because Python is case sensitive ('of' does not equal 'Of')

print(2 in dictionary)
print("Foundation" in dictionary)
# False because dictionary membership checks keys, not values.
```

Operator Precedence and Associativity

***Operator Precedence:***
* It dictates the order in which operations are performed in an expression.
* Operators with higher precedence are evaluated before operators with lower precedence, like multiplication before addition.
* Parentheses can be used to override the default evaluation order.



```python
# multiply(*) will happen first as it has higher precedence and + has lower precedence
expr = 10 + 20 * 30
print(expr)
```

Logical operators (and, or, not) also follow precedence.
* precedence order:  NOT -> AND -> OR
* e.g., I will go if it doesn't(NOT) rain AND I finish work OR have a car. (Follows natural language logic)
([click for revision](https://www.geeksforgeeks.org/python/precedence-and-associativity-of-operators-in-python/))

```python
# Precedence of 'or' & 'and'
name = "Alex"
age = 0
if name == "Alex" or (name == "John" and age >= 2):
    print("Hello! Welcome.")
else:
    print("Good Bye!!")

# name == "John" and age >= 2 is checked first, evaluates to False
# since name == "Alex" is True, the condition passes

# This will evaluate same even if () is removed bcoz it follows the logical operator precedence, i.e AND first, OR after.
# Good practice to write () for better undersanding and readibility of code.
```

***Operators Associativity:***
* It defines the evaluation direction for operators of equal precedence in an expression.
* Most operators evaluate left to right (+, -, *, /, %, AND, OR)
* Exponentiation and assignment group evaluates right to left (**, =, +=, -=, *=, and /=)

```python
exp = 250 / 10 * 10
print(exp)
```
