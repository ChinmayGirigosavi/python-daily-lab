# *Input & Output*

1. Multiple Inputs




```python
#Taking multiple inputs at once by using the split() method
name, middlename, surname = input("Enter your full name: ").split()
print(f"Your full name is {name} {middlename} {surname}")
```

2. Data type and Type Casting (type conversion)

```python
a = 10
b = 2.5
print(type(a), type(b))

#type cast to float
a = float(a)
print(a, "is now a", type(a), "dataype.")

#type cast to str
a = str(a)
print(f"{a} is now a {type(a)} datatype.")
```

3. Type conversion while using input() function

```python
#the default return type of the input() function is string(str)
num = input("Enter a number: ")
print(f'{num} is {type(num)} dataype')

#Using explicit typecasting datatype conversion is possible at the output
num = float(input("Enter a float strictly: "))
print(f'{num} is {type(num)} datatype')
```

5. [Concept of Object Reference](https://www.geeksforgeeks.org/python/python-variables/)

```python
#Variable Reassignment
x = 23
y = x
y = y + 1

print(x)
print(y)

"""Explaination:
In Python, the statement x = 23 means that x is a reference (or pointer)
to an integer object that contains the value 23.
• Python creates an integer object in memory to hold the value 23.
• The variable x is simply a name (or label) that points to that object's memory location.
"""

user_name = "Chinmay"
user_id = user_name
user_id = user_id + " Dhumal"

print(user_name)
print(user_id)
```

7. Swaping variables

```python
# By using multiple assignments it is possible to swap the values of multiple variables without creating a temporary variable
a, b = 19, 23
a, b = b, a
print(a, b)


c, d, e = 5, 15, 25
c, d, e = e, d, c
print(c, d, e)

# c, d, e = e, c, d is also possible (here c=e, d=c, e=d)
c, d, e = e, c, d
print(c, d, e)
```

# [*Operators*](https://www.geeksforgeeks.org/python/python-operators/)


1. Arithmetic Operators

```python
x = 5
y = 2

# the division operator (/) returns a floating-point result.
print("Division: ", x/y)

# while the floor division (//) returns an integer result.
print("Floor Divison: ", x//y)

# Modulus returns remainder
print("Modulus: ", x%y)

# Exponentiation
print("Exponent: ", x**y)
```

2. Comparision Operators

```python
# Comparison(or Relational) operators compares values.
# It either returns True or False according to the condition

x = 22
y = 10

print(x > y)
print(x < y)
print(x == y)
print(x != y)
print(x >= y)
print(x <= y)
```

3. Logical Operators

```python
# Logcial operators are Logical AND, OR, NOT
# And are used to combine conditional statements

x = 22
y = 20
print(x > y and x < y)
print(x > y or x < y)
print(not(x > y))
```

4. Assignment Operators

```python
# Assignment operators are used to assign values to the variables.
# This operator is used to assign the value of the right side of the expression to the left side operand.

a = 20
b = a
print(b)

b += a
print(b)

b -= a
print(b)

b *= a
print(b)

b /= a
print(b)
```
